```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# replay_experimental_0

```python
log_file_path = Path("/YAZILIM/Machine Learning/Deep Learnnig/RL/rl_atari/experiments/agent/replay_experimental_0/logs")
# files start with "metrics_ep"
log_files = list(log_file_path.glob("metrics_ep*.csv"))
log_files.sort(key=lambda x: int(x.stem.split("_")[2]))  # Sort by episode number
log_files
```

```python
# Load the CSV and merge them into a single DataFrame
df = pd.concat([pd.read_csv(file) for file in log_files], ignore_index=True)
df.head()
```

```python
inserts_updates = df[['global_step', 'episode', 'total_dnd_inserts', 'total_dnd_updates']].copy()
inserts_updates.drop_duplicates(inplace=True)
inserts_updates.reset_index(drop=True, inplace=True)
inserts_updates.head()
```

```python
inserts_updates['global_step'].diff() 
```

```python
inserts_updates['transition_count'] = (inserts_updates['global_step'].diff() - 174).fillna(0).astype(int) 
```

```python
inserts_updates.head()
```

```python
inserts_updates['insert_rate'] = np.divide(inserts_updates['total_dnd_inserts'], inserts_updates['transition_count'])    
inserts_updates.head()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(inserts_updates['episode'], inserts_updates['insert_rate'], marker='o', linestyle='-', color='b')
plt.title('Insert Rate Over Episodes')
```

# replay_experimental_2

```python
log_file_50 = Path("/YAZILIM/Machine Learning/Deep Learnnig/RL/rl_atari/experiments/agent/replay_experimental_2/logs/metrics_ep_50_opt_step_0_6749.csv")
df = pd.read_csv(log_file_50)
df.head()
```

```python
inserts_updates = df[['global_step', 'episode', 'total_dnd_inserts', 'total_dnd_updates']].copy()
inserts_updates.drop_duplicates(inplace=True)
inserts_updates['transition_count'] = (inserts_updates['global_step'].diff() - 174).fillna(0).astype(int) 
inserts_updates['insert_rate'] = np.divide(inserts_updates['total_dnd_inserts'], inserts_updates['transition_count'])    
inserts_updates.head()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(inserts_updates['episode'], inserts_updates['insert_rate'], marker='o', linestyle='-', color='b')
plt.title('Insert Rate Over Episodes for Episode 50')
```

```python
def calculate_insert_rates_manually(steps_and_inserts):
    rates = []
    for global_st_lag1, global_st, total_dnd_inserts in steps_and_inserts:
        transition_count = global_st - global_st_lag1 - 174
        insert_rate = total_dnd_inserts / transition_count if transition_count > 0 else 0
        rates.append(insert_rate)
    return rates
```

```python
steps_and_inserts = [(39116, 39721, 419),
                     (39721, 40460, 544),
                     (40460, 41158, 505),
                     (41158, 41555, 201),
                     (41555, 42486, 730),
                     (42486, 43203, 506),
                     (43203, 44034, 627),
                     (44034, 44622, 393),
                     (44622, 45031, 225),
                     (45031, 45799, 575)]
rates = calculate_insert_rates_manually(steps_and_inserts)
```

```python
plt.plot(range(len(rates)), rates, marker='o', linestyle='-', color='b')  # Plot the insert rates
plt.show()
```
